```python
import pandas as pd
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import classification_report, confusion_matrix
from tensorflow.keras import layers, constraints
from tensorflow.keras.utils import to_categorical

# Load ECG data
X = np.load('X.npy')  # shape: (samples, features)
y = np.load('y.npy')  # shape: (samples,)

# Optional: If X has more than 2 dimensions, flatten it
if len(X.shape) > 2:
    X = X.reshape(X.shape[0], -1)

# Standardize features
scaler = StandardScaler()
X = scaler.fit_transform(X)

# Encode labels to categorical
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
y_categorical = to_categorical(y_encoded)

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y_categorical, test_size=0.2, random_state=42, stratify=y_categorical)

num_features = X.shape[1]
num_classes = y_categorical.shape[1]

print("Dataset shapes:")
print(f"X_train shape: {X_train.shape}, X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}, y_test shape: {y_test.shape}")
print(f"Number of features: {num_features}, Number of classes: {num_classes}")

# ----------------- Custom Layers ----------------- #


class WeightedFeatureLayer(layers.Layer):
    def __init__(self, **kwargs):
        super().__init__(**kwargs)

    def build(self, input_shape):
        self.w_x = self.add_weight(
            shape=(input_shape[-1],), initializer="ones", trainable=False, name="w_x"
        )
        self.w_sin = self.add_weight(
            shape=(input_shape[-1],), initializer="ones", trainable=True, name="w_sin"
        )
        self.w_cos = self.add_weight(
            shape=(input_shape[-1],), initializer="ones", trainable=True, name="w_cos"
        )

    def call(self, inputs):
        transformed = tf.concat([
            self.w_x * inputs,
            self.w_sin * tf.sin(inputs),
            self.w_cos * tf.cos(inputs)
        ], axis=-1)
        return transformed

class WeightedFeatureLayer5(layers.Layer):
    def __init__(self, **kwargs):
        super().__init__(**kwargs)

    def build(self, input_shape):
        self.w_x = self.add_weight(
            shape=(input_shape[-1],), initializer="ones", trainable=False, name="w_x"
        )
        self.w_sin = self.add_weight(
            shape=(input_shape[-1],), initializer="ones", trainable=True, name="w_sin"
        )
        self.w_cos = self.add_weight(
            shape=(input_shape[-1],), initializer="ones", trainable=True, name="w_cos"
        )
        self.w_sin2x = self.add_weight(
            shape=(input_shape[-1],), initializer="ones", trainable=True, name="w_sin2x"
        )
        self.w_cos2x = self.add_weight(
            shape=(input_shape[-1],), initializer="ones", trainable=True, name="w_cos2x"
        )

    def call(self, inputs):
        transformed = tf.concat([
            self.w_x * inputs,
            self.w_sin * tf.sin(inputs),
            self.w_cos * tf.cos(inputs),
            self.w_sin2x * tf.sin(2 * inputs),
            self.w_cos2x * tf.cos(2 * inputs)
        ], axis=-1)
        return transformed

# ----------------- Model(FLANN) ----------------- #

def create_ecg_multiclass_model(input_shape, num_classes):
    model = tf.keras.Sequential([
        layers.Input(shape=(input_shape,)),
        WeightedFeatureLayer(),
        layers.Dense(num_classes, activation='softmax',trainable=False)  # For multi-class
    ])
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss='categorical_crossentropy',  # Multi-class loss
        metrics=['accuracy', tf.keras.metrics.Precision(name='precision'), tf.keras.metrics.Recall(name='recall')]
    )
    return model

def create_ecg_multiclass_model_5exp(input_shape, num_classes):
    model = tf.keras.Sequential([
        layers.Input(shape=(input_shape,)),
        WeightedFeatureLayer5(),
        layers.Dense(num_classes, activation='softmax', trainable=False)  # For multi-class
    ])
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss='categorical_crossentropy',  # Multi-class loss
        metrics=['accuracy', tf.keras.metrics.Precision(name='precision'), tf.keras.metrics.Recall(name='recall')]
    )
    return model

# 3-Expansion Model
print("=" * 50)
print("3-Expansion FLANN Model")
print("=" * 50)
model_3exp = create_ecg_multiclass_model(num_features, num_classes)
model_3exp.summary()



# ----------------- Training 3-Expansion Model ----------------- #

print("\n" + "=" * 50)
print("Training 3-Expansion FLANN Model")
print("=" * 50)
history3 = model_3exp.fit(
    X_train.astype(np.float32),
    y_train,
    validation_split=0.2,
    epochs=100,
    batch_size=1,
    callbacks=[
        tf.keras.callbacks.EarlyStopping(patience=300, restore_best_weights=True)
    ],
    verbose=1
)

# ----------------- Evaluation 3-Expansion Model ----------------- #

print("\n" + "=" * 50)
print("Evaluation: 3-Expansion FLANN Model")
print("=" * 50)
test_preds_3exp = model_3exp.predict(X_test.astype(np.float32))
predicted_labels_3exp = np.argmax(test_preds_3exp, axis=1)
true_labels = np.argmax(y_test, axis=1)

print("\nClassification Report (3-Expansion):")
print(classification_report(true_labels, predicted_labels_3exp, target_names=label_encoder.classes_.astype(str)))

print("\nConfusion Matrix (3-Expansion):")
print(confusion_matrix(true_labels, predicted_labels_3exp))

# Calculate metrics for 3-expansion
test_loss_3exp, test_accuracy_3exp, test_precision_3exp, test_recall_3exp = model_3exp.evaluate(
    X_test.astype(np.float32), y_test, verbose=0
)
```

```python
# ----------------- 5-Expansion Model ----------------- #

print("\n" + "=" * 50)
print("5-Expansion FLANN Model")
print("=" * 50)
model_5exp = create_ecg_multiclass_model_5exp(num_features, num_classes)
model_5exp.summary()
```

```python
# ----------------- Training 5-Expansion Model ----------------- #

print("\n" + "=" * 50)
print("Training 5-Expansion FLANN Model")
print("=" * 50)
history5 = model_5exp.fit(
    X_train.astype(np.float32),
    y_train,
    validation_split=0.2,
    epochs=100,
    batch_size=1,
    callbacks=[
        tf.keras.callbacks.EarlyStopping(patience=300, restore_best_weights=True)
    ],
    verbose=1
)
```

```python
# ----------------- Evaluation 5-Expansion Model ----------------- #

print("\n" + "=" * 50)
print("Evaluation: 5-Expansion FLANN Model")
print("=" * 50)
test_preds_5exp = model_5exp.predict(X_test.astype(np.float32))
predicted_labels_5exp = np.argmax(test_preds_5exp, axis=1)

print("\nClassification Report (5-Expansion):")
print(classification_report(true_labels, predicted_labels_5exp, target_names=label_encoder.classes_.astype(str)))

print("\nConfusion Matrix (5-Expansion):")
print(confusion_matrix(true_labels, predicted_labels_5exp))

# Calculate metrics for 5-expansion
test_loss_5exp, test_accuracy_5exp, test_precision_5exp, test_recall_5exp = model_5exp.evaluate(
    X_test.astype(np.float32), y_test, verbose=0
)
```

```python
# ----------------- Comparison of 3-Expansion vs 5-Expansion ----------------- #

print("\n" + "=" * 70)
print("COMPARISON: 3-Expansion vs 5-Expansion FLANN")
print("=" * 70)

# Create comparison table
comparison_data = {
    'Metric': ['Test Accuracy', 'Test Loss', 'Test Precision', 'Test Recall'],
    '3-Expansion': [
        f"{test_accuracy_3exp:.4f}",
        f"{test_loss_3exp:.4f}",
        f"{test_precision_3exp:.4f}",
        f"{test_recall_3exp:.4f}"
    ],
    '5-Expansion': [
        f"{test_accuracy_5exp:.4f}",
        f"{test_loss_5exp:.4f}",
        f"{test_precision_5exp:.4f}",
        f"{test_recall_5exp:.4f}"
    ]
}

comparison_df = pd.DataFrame(comparison_data)
print("\nPerformance Metrics Comparison:")
print(comparison_df.to_string(index=False))

# Calculate improvement
accuracy_improvement = test_accuracy_5exp - test_accuracy_3exp
loss_improvement = test_loss_3exp - test_loss_5exp  # Lower is better
precision_improvement = test_precision_5exp - test_precision_3exp
recall_improvement = test_recall_5exp - test_recall_3exp

print("\n" + "-" * 70)
print("Improvement Analysis:")
print("-" * 70)
print(f"Accuracy Improvement: {accuracy_improvement:+.4f} ({accuracy_improvement*100:+.2f}%)")
print(f"Loss Improvement: {loss_improvement:+.4f} ({'Better' if loss_improvement > 0 else 'Worse'})")
print(f"Precision Improvement: {precision_improvement:+.4f} ({precision_improvement*100:+.2f}%)")
print(f"Recall Improvement: {recall_improvement:+.4f} ({recall_improvement*100:+.2f}%)")

# Model complexity comparison
print("\n" + "-" * 70)
print("Model Complexity Comparison:")
print("-" * 70)
print(f"3-Expansion: {num_features} input features → {num_features * 3} expanded features")
print(f"5-Expansion: {num_features} input features → {num_features * 5} expanded features")
print(f"Feature expansion ratio: 3x vs 5x")
print(f"Total trainable parameters (3-exp): {model_3exp.count_params()}")
print(f"Total trainable parameters (5-exp): {model_5exp.count_params()}")

# Determine which model performs better
print("\n" + "=" * 70)
print("CONCLUSION:")
print("=" * 70)
if test_accuracy_5exp > test_accuracy_3exp:
    print(f"✓ 5-Expansion model performs BETTER with {test_accuracy_5exp:.4f} accuracy vs {test_accuracy_3exp:.4f}")
else:
    print(f"✓ 3-Expansion model performs BETTER with {test_accuracy_3exp:.4f} accuracy vs {test_accuracy_5exp:.4f}")

if test_loss_5exp < test_loss_3exp:
    print(f"✓ 5-Expansion model has LOWER loss: {test_loss_5exp:.4f} vs {test_loss_3exp:.4f}")
else:
    print(f"✓ 3-Expansion model has LOWER loss: {test_loss_3exp:.4f} vs {test_loss_5exp:.4f}")
```
